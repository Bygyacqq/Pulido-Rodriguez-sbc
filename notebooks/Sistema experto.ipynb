# Introducción

Los modelos de *machine learning* permiten el análisis de datos, la identificación de patrones y la generación de predicciones que apoyan la toma de decisiones en diversos contextos. Al momento de elegir un modelo de *machine learning*, es necesario tener en cuenta factores clave como el objetivo del análisis, las características de los datos disponibles, las relaciones entre las variables y la necesidad de interpretar los resultados. Si se selecciona un modelo sin considerar estos factores, las recomendaciones pueden no ajustarse a las necesidades del problema.

Teniendo esto en cuenta, se propone el desarrollo de un sistema experto que permita orientar la selección de modelos de *machine learning* mediante reglas de conocimiento definidas previamente. El sistema analizará las características del problema proporcionadas por el usuario y generará recomendaciones justificadas, indicando qué modelos podrían ser adecuados para abordar cada necesidad específica.

Esta propuesta contempla cuatro objetivos de análisis: clasificación, regresión, agrupamiento (*clustering*) y detección de anomalías. Su finalidad es representar parte del conocimiento necesario para orientar una decisión inicial, sin reemplazar la experimentación ni la evaluación técnica de los resultados.

# Definición del dominio

El dominio seleccionado es la recomendación de modelos de *machine learning* según el objetivo del análisis y las características de los datos disponibles.

El sistema considerará cuatro tipos de problemas:

- **Clasificación:** Cuando el objetivo sea predecir una categoría, como determinar si un cliente responderá o no a una campaña.
- **Regresión:** Cuando el objetivo sea estimar un valor numérico, como la cantidad de unidades que podría venderse de un producto.
- **Agrupamiento (Clustering):** Cuando se busque organizar los datos en grupos según sus características y similitudes, como segmentar clientes de acuerdo con sus comportamientos.
- **Detección de anomalías:** Cuando se busque identificar observaciones que presenten comportamientos inusuales respecto al conjunto de datos, como detectar transacciones con características atípicas.

A partir de estas características identificadas, el sistema aplicará las reglas de conocimiento correspondientes para recomendar modelos potencialmente apropiados para cada situación.

# Justificación

La selección de un modelo predictivo constituye una decisión importante dentro de un proyecto de análisis de datos, ya que no todos los problemas pueden abordarse mediante un mismo enfoque. Un modelo adecuado para predecir categorías no necesariamente será apropiado para estimar valores numéricos, identificar agrupaciones o detectar comportamientos anómalos.

El desarrollo de este sistema permitirá representar conocimientos relacionados con la selección inicial de modelos mediante lógica de predicados y reglas de producción. Asimismo, facilitará la comprensión de los procesos de razonamiento de un sistema experto mediante la identificación de hechos, la aplicación de reglas y la generación de conclusiones explicables.

# Alcance y limitaciones

Este sistema recibirá información general sobre el objetivo del análisis y las características de los datos disponibles. A partir de esta información, aplicará una base de conocimiento para recomendar modelos de clasificación, regresión, agrupamiento (*clustering*) y detección de anomalías, explicando las condiciones que justifican sus recomendaciones.

El alcance se limita a la recomendación basada en reglas previamente definidas. El sistema no entrenará modelos, no utilizará datos reales de los clientes ni garantizará que el modelo recomendado obtenga el mejor desempeño en un problema específico.

La validación de las recomendaciones se realizará mediante casos de prueba diseñados para comprobar el correcto funcionamiento de las reglas, el motor de inferencia y el mecanismo de explicación de las conclusiones.

# Formalización de las reglas en lógica de predicados

En las expresiones lógicas, la variable `x` representa un problema de análisis de datos, la variable `t` representa el objetivo del problema, la variable `e` representa el estado de etiquetado, la variable `r` representa el tipo de relación entre las variables, la variable `g` representa la estructura de los grupos y la variable `m` representa un modelo de *machine learning*. El cuantificador universal `∀x` indica que cada regla se aplica a cualquier problema que cumpla las condiciones establecidas.

### Regla 1. Regresión logística

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clasificacion) ∧ DatosTabulares(x) → RecomendarModelo(x,RegresionLogistica)]
</div>

- **Interpretación:** Si se busca clasificar información y los datos son tabulares, se recomienda la regresión logística.
- **Justificación:** Es una alternativa para problemas de clasificación y permite estimar probabilidades asociadas a las categorías.

### Regla 2. Árbol de decisión para clasificación

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clasificacion) ∧ RequiereInterpretabilidad(x) → RecomendarModelo(x,ArbolDecisionClasificacion)]
</div>

- **Interpretación:** Si el objetivo es la clasificación y se requiere interpretar fácilmente las decisiones, se recomienda el uso de un árbol de decisión.
- **Justificación:** Sus ramas y condiciones permiten explicar cómo se generan las predicciones.

### Regla 3. Random Forest para clasificación

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clasificacion) ∧ Relaciones(x,Complejas) → RecomendarModelo(x,RandomForestClasificacion)]
</div>

- **Interpretación:** Si el objetivo es la clasificación y existen relaciones complejas entre las variables, se recomienda Random Forest.
- **Justificación:** Combina múltiples árboles de decisión y puede identificar patrones no lineales en los datos.

### Regla 4. Gradient Boosting para clasificación

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clasificacion) ∧ Relaciones(x,Complejas) ∧ DatosTabulares(x) → RecomendarModelo(x,GradientBoostingClasificacion)]
</div>

- **Interpretación:** Si el problema es de clasificación, existen relaciones complejas y los datos son tabulares, se recomienda Gradient Boosting.
- **Justificación:** Este modelo construye secuencialmente modelos que corrigen errores anteriores y permite identificar patrones complejos.

### Regla 5. Clasificación con probabilidades

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clasificacion) ∧ RequiereProbabilidades(x) → RecomendarModelo(x,RegresionLogistica)]
</div>

- **Interpretación:** Si el problema es de clasificación y se necesitan probabilidades asociadas a las categorías, se recomienda regresión logística.
- **Justificación:** Este modelo permite estimar probabilidades para las diferentes clases y facilita la interpretación de los resultados.

### Regla 6. Regresión lineal

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Regresion) ∧ Relaciones(x,Lineales) → RecomendarModelo(x,RegresionLineal)]
</div>

- **Interpretación:** Si se busca estimar un valor numérico y las relaciones entre las variables son aproximadamente lineales, se recomienda regresión lineal.
- **Justificación:** Es una alternativa inicial para estimar valores cuando una relación lineal resulta razonable.

### Regla 7. Árbol de decisión para regresión

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Regresion) ∧ RequiereInterpretabilidad(x) → RecomendarModelo(x,ArbolDecisionRegresion)]
</div>

- **Interpretación:** Si se quiere predecir un valor numérico y se requiere interpretar las decisiones, se recomienda un árbol de decisión para regresión.
- **Justificación:** Permite representar relaciones mediante condiciones y seguir el proceso que conduce a una predicción.

### Regla 8. Random Forest para regresión

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Regresion) ∧ Relaciones(x,Complejas) ∧ DatosTabulares(x) → RecomendarModelo(x,RandomForestRegresion)]
</div>

- **Interpretación:** Si se busca predecir un valor numérico con datos tabulares y relaciones complejas, se recomienda Random Forest para regresión.
- **Justificación:** Combina árboles de decisión y puede modelar relaciones complejas entre las variables.

### Regla 9. Gradient Boosting para regresión

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Regresion) ∧ Relaciones(x,Complejas) ∧ DatosTabulares(x) → RecomendarModelo(x,GradientBoostingRegresion)]
</div>

- **Interpretación:** Si el problema es de regresión, existen relaciones complejas y los datos son tabulares, se recomienda Gradient Boosting para regresión.
- **Justificación:** Este modelo permite corregir progresivamente los errores de predicción y capturar relaciones complejas entre las variables.

### Regla 10. K-Means para clustering

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clustering) ∧ EstructuraGrupos(x,Compactos) ∧ DatosTabulares(x) → RecomendarModelo(x,KMeans)]
</div>

- **Interpretación:** Si el objetivo es agrupar datos, se identifican grupos compactos y los datos son tabulares, se recomienda K-Means.
- **Justificación:** Este algoritmo organiza los datos en un número determinado de grupos, buscando que los elementos de cada grupo sean similares entre sí.

### Regla 11. DBSCAN para clustering

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Clustering) ∧ EstructuraGrupos(x,DensidadesVariables) → RecomendarModelo(x,DBSCAN)]
</div>

- **Interpretación:** Si el objetivo es agrupar datos y los grupos presentan densidades variables, se recomienda DBSCAN.
- **Justificación:** Este algoritmo identifica agrupaciones a partir de la densidad de los datos y permite detectar observaciones que no pertenecen a ningún grupo.

### Regla 12. Isolation Forest para detección de anomalías

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Anomalias) ∧ DatosTabulares(x) → RecomendarModelo(x,IsolationForest)]
</div>

- **Interpretación:** Si el objetivo es identificar observaciones anómalas y los datos son tabulares, se recomienda Isolation Forest.
- **Justificación:** Este algoritmo identifica observaciones atípicas mediante procesos de aislamiento y permite detectar registros que presentan comportamientos diferentes al resto.

### Regla 13. One-Class SVM para detección de anomalías

**Expresión lógica:**

<div align="center">
∀x [Objetivo(x,Anomalias) ∧ DatosEscalados(x) → RecomendarModelo(x,OneClassSVM)]
</div>

- **Interpretación:** Si el objetivo es detectar anomalías y las variables numéricas se encuentran escaladas, se recomienda One-Class SVM.
- **Justificación:** Este algoritmo aprende las características generales de los datos considerados normales y permite identificar observaciones que se alejan de ese comportamiento.

### Regla 14. Identificación del aprendizaje supervisado

**Expresión lógica:**

<div align="center">
∀x [RequiereEtiquetas(x) ∧ EstadoEtiquetas(x,Etiquetados) → ModeloSupervisado(x)]
</div>

- **Interpretación:** Si el problema requiere datos etiquetados y estos se encuentran disponibles, se identifica que puede abordarse mediante aprendizaje supervisado.
- **Justificación:** Los modelos supervisados aprenden a partir de ejemplos que incluyen respuestas o valores conocidos para realizar predicciones.

### Regla 15. Advertencia por ausencia de datos etiquetados

**Expresión lógica:**

<div align="center">
∀x [RequiereEtiquetas(x) ∧ EstadoEtiquetas(x,NoEtiquetados) → AdvertenciaDatosNoEtiquetados(x)]
</div>

- **Interpretación:** Si el problema requiere datos etiquetados, pero estos no están disponibles, el sistema genera una advertencia sobre esta condición.
- **Justificación:** Los modelos supervisados requieren datos etiquetados para su entrenamiento. La regla informa al usuario sobre esta limitación sin impedir que se consideren otras alternativas.

La base de conocimiento reúne las reglas de producción utilizadas por el sistema experto para recomendar modelos de *machine learning*. Cada regla contiene un identificador, un conjunto de condiciones, una conclusión y una descripción que facilita la comprensión de su propósito.

Para esta implementación, se utilizará una lista en Python que almacenará las 15 reglas previamente formalizadas en este notebook. Estas reglas estarán organizadas según los objetivos de clasificación, regresión, clustering y detección de anomalías, además de incluir condiciones relacionadas con la disponibilidad de datos etiquetados.

La función `crear_base_conocimiento()` será la responsable de construir y devolver la lista de reglas que utilizará el sistema durante el proceso de encadenamiento hacia adelante.


```python
def crear_regla(identificador, condiciones, conclusion, descripcion):
    """
    Crea una regla de producción para el sistema:
    
    Parámetros:
    - identificador: Nombre o número de la regla.
    - condiciones: Hechos los cuáles se deben cumplir.
    - conclusiones: Hecho que se obtiene al aplicar la regla.
    - descripcion: Explicación del funcionamiento de la regla.

    Retorna:
    Un diccionario representativo de la regla.
    
    """
    return {
        "id": identificador,
        "condiciones": condiciones,
        "conclusion": conclusion,
        "descripcion": descripcion
    }
```

# Encadenamiento hacia delante
Es un mecanismo de razonamiento que parte de conjunto de hechos conocidos y utiliza las reglas de base de conocimiento para obtener nuevas conclusiones.

El sistema compara las condiciones de cada regla con los hechos disponibles. Cuándo todas las condiciones de una regla se cumplen, su conclusión se agrega al conjunto de hechos y registra la regla aplicada.

El proceso se repite hasta que no se puedan generar nuevas conclusiones. De esta forma, el sistema puede identificar modelos que se ajusten a las características del problema planteado.

Para la implementación de este mecanismo será utilizado en una estructura iterativa que recorrerá las reglas y verificará sus condiciones hasta alcanzar un estado en el que no se produzcan nuevos hechos.

```python
def encadenamiento_hacia_adelante(hechos_iniciales, base_conocimiento, max_iteraciones=100):
    """
    Aplica encadenamiento hacia adelante para poder obtener nuevas conclusiones a partir de los hechos inciales.

    Parámetros:
    hechos_inciales: Conjunto de hechos que ya se conocen.
    base_conocimineto: Lista de reglas de producción.

    Retorna:
    hechos: Cojunto de hechos inciales y derivados.
    reglas_aplicadas: Lista de las reglas que fueron utilizadas.
    
    """

    hechos = set(hechos_iniciales)
    reglas_aplicadas = []
    hubo_cambios = True
    iteracion = 0
    estados_visitados = set()
    logging.info("Se está iniciando el proceso de inferencia")


    while hubo_cambios:
        if iteracion >= max_iteraciones:
            logging.warning("El limite de iteraciones permitidas se ha alcanzado.")
            break
        estado_actual = frozenset(hechos)
        if estado_actual in estados_visitados:
            logging.warning("Existen estados repetidos. El proceso se detuvo por seguridad.")
            break
        estados_visitados.add(estado_actual)
        iteracion += 1
        hubo_cambios = False
        logging.info(f"Iteración {iteracion}.")
        for regla in base_conocimiento:
            condiciones = set(regla["condiciones"])
            conclusion = regla["conclusion"]
            if condiciones.issubset(hechos):
                if regla not in reglas_aplicadas:
                    reglas_aplicadas.append(regla)
                    logging.info(f"Regla activada: {regla['id']} ")
                if conclusion not in hechos:
                    hechos.add(conclusion)
                    hubo_cambios = True
                    logging.info(f"Nueva conclusión: {conclusion}")
    
    logging.info("Proceso de inferencia finalizado.")

    return hechos, reglas_aplicadas
```

# Función de explicación de conclusiones

La función `porque()` permite la consulta del por qué el sistema llegó a la conclusión. Para esto, busca entre las reglas aplicadas aquellas cuya conclusión coincida con la que desea consultar.
Esta función facilita la interpretación del sistema, ya que permite identificar la regla que generó una recomendación y conocer las condiciones que llevaron a obtenerla. 

```python
def porque(conclusion, reglas_aplicadas):
    """
    Explica el por qué se obtuvo al conclusión.
    """
    reglas_encontradas = [
        regla for regla in reglas_aplicadas
        if regla["conclusion"] == conclusion
    ]
    if not reglas_encontradas:
        print("No se encontró una regla aplicada la cuál genere esta conclusión.")
        return

    print("Consulta de la explicación")
    print("-" * 40)
    print(f"Conclusión consultada: {conclusion}")

    for regla in reglas_encontradas:
        print(f"\nRegla responsable: {regla['id']}")
        print(f"Descripción:{regla['descripcion']}")
        print(f"Condiciones necesarias:")

        for condicion in regla["condiciones"]:
            print(f"- {condicion}")
            
    print("\nFin de la explicación")
```
